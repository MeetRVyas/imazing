# 5. Filtering & Enhancement


### Blurring & Sharpening


```python
img = Imazing("lena.jpg")

# --- Basic: Gaussian Blur ---
img.blur(method="gaussian", ksize=15)

# --- Advanced Variations ---

# 1. Median Blur (Good for salt-and-pepper noise)
# img.blur(method="median", ksize=5)

# 2. Sharpen Image (Enhance edges)
# img = Imazing("lena.jpg").sharpen()

# 3. Denoise (Computationally expensive but good results)
# img = Imazing("lena.jpg").denoise(strength=15)

plt.imshow(cv2.cvtColor(img.to_numpy(), cv2.COLOR_BGR2RGB))
plt.title("Filtered")
plt.show()
```

### Edge Detection & Thresholding


```python
img = Imazing("lena.jpg")

# --- Basic: Canny Edge Detection ---
img.detect_edges(method='canny', t1=100, t2=200)

# --- Advanced Variations ---

# 1. Sobel Gradient Detection
# img = Imazing("lena.jpg").detect_edges(method='sobel')

# 2. Otsu Thresholding (Binarization)
# img = Imazing("lena.jpg").segment_threshold(type='otsu')

# 3. Adaptive Thresholding (For uneven lighting)
# img = Imazing("lena.jpg").segment_threshold(type='adaptive')

plt.imshow(img.to_numpy(), cmap='gray')
plt.title("Edges / Segments")
plt.show()
```

# 6. Features, Detection & AI


### Face Detection


```python
img = Imazing("lena.jpg")

# --- Basic: Detect Faces ---
faces = img.detect_faces()

print(f"Faces detected: {len(faces)}")

# Draw rectangles around faces
for (x, y, w, h) in faces:
    img.draw_rect(x, y, w, h, color=(0, 255, 0), thickness=2)
    img.draw_text("Person", x, y-10)

# --- Advanced Variations ---

# 1. Crop and Save Faces
# for i, (x, y, w, h) in enumerate(faces):
#     face_crop = Imazing(img.to_numpy()[y:y+h, x:x+w])
#     face_crop.save(f"face_{i}.jpg")

plt.imshow(cv2.cvtColor(img.to_numpy(), cv2.COLOR_BGR2RGB))
plt.show()
```
